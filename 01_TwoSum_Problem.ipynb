# Two Sum

You are given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

 

Example 1:

Input: nums = [2,7,11,15], target = 9

Output: [0,1]

Explanation:

The sum of nums[0] and nums[1] is 2 + 7 = 9, which equals the target. Therefore, the indices [0, 1] are returned.

Example 2:

Input: nums = [3,2,4], target = 6

Output: [1,2]

Explanation:

The sum of nums[1] and nums[2] is 2 + 4 = 6, which equals the target. Therefore, the indices [1, 2] are returned.

Example 3:

Input: nums = [3,3], target = 6

Output: [0,1]

Explanation:

The sum of nums[0] and nums[1] is 3 + 3 = 6, which equals the target. Therefore, the indices [0, 1] are returned.

 

Constraints:

2 <= nums.length <= 104
-109 <= nums[i] <= 109
-109 <= target <= 109
Only one valid answer exists.
 

Follow-up: Can you come up with an algorithm that is less than O(n2) time complexity?

# V1: Brute force approach 
Nested for loops that check all the possibilites resulting in a O(n**2) complexity

```python
def two_sums_v1(nums: list[int], target: int) -> list[int]:
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i,j]
            
    raise ValueError("No two numbers sum to the target")
    
```

#  Optimised approach 

Using a dictionary to check and search already seen numbers reduces the complexity to O(n)

## V2: Use the fill first check alter approach for the dictionary

```python
def two_sums_v2(nums: list[int], target: int) -> list[int]:
    seen= {}
    for i, x in enumerate(nums): #fill
        seen[x] = i
    for i, x in enumerate(nums):
        need = target - x
        if need in seen and seen[need] !=i:
            return [seen[need], i]
        
    raise ValueError("No two numbers sum to the target")

```

## V3: Use the fill as you check approach

```python
def two_sums_v3(nums: list[int], target: int) -> list[int]:
    seen= {}
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
       
        seen[x] = i

    raise ValueError("No two numbers sum to the target")    
```

## Further Optimisation
Further otpimisation can be achieved by utilising preprocessing. In this approach, if a large list of numbers is used as a input we can use a pre-processing pass to index the dictionary. This still yields a computation that is O(n), however, there is a modest computational cost saving for every subsequent query is which is only O(1) instead of O(n) on each run

First we build the separate funcions

```python
def build_index(nums: list[int]) -> dict[int, list[int]]:
    """ Function reads a list of numbers (integers) and indexes them into a dictionary.
    Special care is taken to return a list of indicies for each number in order to tackle the problem 
    with overwriting existing values if there a repeating numbers in the list"""
    index = {}
    for i, x in enumerate(nums):
        index.setdefault(x, []).append(i) 
    return index

def query(index: dict[int, list[int]], target: int) -> list[int]:
    for x, position in index.items():
        need = target - x
        if need == x:
            if len(position) >= 2:
                return[position[0], position[1]]
        elif need in index:
            return[position[0], index[need][0]]
          
```

Now create a class that utilises them

```python
class TwoSum:
    def __init__(self, nums: list[int]):
        self._index = build_index(nums)

    def find(self, target: int) -> list[int]:
        return query(self._index, target)
```

Example

```python
finder = TwoSum(nums)
finder.find(9)
```

```python
finder = TwoSum([3, 2, 4]) 
finder.find(6) 
```
